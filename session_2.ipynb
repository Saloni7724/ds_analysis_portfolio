```python
import numpy as np

np.random.seed(42)

units_sold = np.random.randint(5, 60, size=(6, 7))

month_units = np.random.randint(5, 60, size=(4, 6, 7))

print("Units Sold:")
print(units_sold)

print("Month Units:")
print(month_units)
```

```python
"""Task 1
"""
high_mask = units_sold > 40

print("True Count:", np.sum(high_mask))

print("Matching Values:")
print(units_sold[high_mask])

top_items = units_sold[[0, 3, 5]]

print("Rice, Tea, Dal:")
print(top_items)

stock_change = units_sold - 30

stock_change = stock_change.copy()

mask = stock_change < 0

print("Values Changed:", np.sum(mask))

stock_change[mask] = 0

print("Updated Stock:")
print(stock_change)
```

```python
"""Task 2

"""
price_col = np.array([
    [52],
    [44],
    [45],
    [380],
    [140],
    [120]
])

revenue_inr = units_sold * price_col

revenue_gst = revenue_inr * 1.05

day_factor = np.array([1, 1, 1, 1, 1, 1.3, 1.4])

weekend_revenue = revenue_inr * day_factor

print("Units Shape:", units_sold.shape)
print("Price Shape:", price_col.shape)

print("Revenue INR:")
print(revenue_inr)

print("Revenue GST:")
print(revenue_gst)

print("Weekend Revenue:")
print(weekend_revenue)

print("Check Row 1:")
print(np.array_equal(revenue_inr[1], units_sold[1] * 44))
```

```python
"""Task 3
"""
row_min = units_sold.min(axis=1, keepdims=True)

row_max = units_sold.max(axis=1, keepdims=True)

units_scaled = (units_sold - row_min) / (row_max - row_min)

print("Scaled Units:")
print(units_scaled)

print("Minimum Check:")
print(np.allclose(units_scaled.min(axis=1), 0))

print("Maximum Check:")
print(np.allclose(units_scaled.max(axis=1), 1))
```

```python
"""Task 4

"""

tea_week = units_sold[3, :].copy()


day3_sales = units_sold[:, 2].copy()

staple_block = units_sold[:3, -4:].copy()

alt_days = units_sold[:, ::2].copy()

dal_last = units_sold[-1, :].copy()


w3_tea_d6 = month_units[2, 3, 5]

print("Tea Week:")
print(tea_week)

print("Day 3 Sales:")
print(day3_sales)

print("Staple Block:")
print(staple_block)

print("Alternate Days:")
print(alt_days)

print("Dal Last:")
print(dal_last)

print("Week 3 Tea Day 6:")
print(w3_tea_d6)
```
