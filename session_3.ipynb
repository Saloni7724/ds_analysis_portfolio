```python
import numpy as np

np.random.seed(42)

units_sold = np.random.randint(5, 60, size=(6, 7))

price_inr = np.array([52, 44, 45, 380, 140, 120])

print("Units Sold:")
print(units_sold)

print("Price:")
print(price_inr)
```

```python
"""Task 1
"""
switch_p = np.array([
    [0.7, 0.2, 0.1],
    [0.3, 0.5, 0.2],
    [0.2, 0.3, 0.5]
])

print("Switching Matrix:")
print(switch_p)

print("Determinant:")
print(np.linalg.det(switch_p))

inverse = np.linalg.inv(switch_p)

print("Inverse Check:")
print(np.allclose(inverse @ switch_p, np.eye(3)))

vals, vecs = np.linalg.eig(switch_p.T)

index = np.argmin(np.abs(vals - 1))

steady_share = vecs[:, index].real

steady_share = steady_share / steady_share.sum()

print("Eigenvalues:")
print(vals)

print("Steady Share:")
print(steady_share)

print("Total Share:")
print(steady_share.sum())
```

```python
"""Task 2
"""
week2 = np.random.randint(5, 60, size=(6, 7))

two_weeks = np.concatenate((units_sold, week2), axis=1)

stacked = np.stack((units_sold, week2))

print("Two Weeks:")
print(two_weeks)

print("Stacked:")
print(stacked)

first_week, second_week = np.split(two_weeks, 2, axis=1)

print("First Week Match:")
print(np.array_equal(first_week, units_sold))

print("Second Week Match:")
print(np.array_equal(second_week, week2))

# Ravel
ravel_arr = two_weeks.ravel()

ravel_arr[0] = 999

print("Ravel Shares Memory:")
print(np.shares_memory(two_weeks, ravel_arr))

# Flatten
flatten_arr = two_weeks.flatten()

flatten_arr[1] = 888

print("Flatten Shares Memory:")
print(np.shares_memory(two_weeks, flatten_arr))
```

```python
""" Task 3
"""
daily1 = price_inr.dot(units_sold)

daily2 = np.matmul(price_inr, units_sold)

daily3 = (units_sold * price_inr.reshape(6, 1)).sum(axis=0)

print("Revenue using dot:")
print(daily1)

print("Revenue using matmul:")
print(daily2)

print("Revenue using multiplication:")
print(daily3)

print("All Results Match:")
print(np.allclose(daily1, daily2) and np.allclose(daily2, daily3))

item_cosales = units_sold @ units_sold.T

print("Item Co-sales:")
print(item_cosales)

print("Shape:")
print(item_cosales.shape)
```

```python
"""Task 4
"""
total = np.sum(units_sold, axis=1)

mean = np.mean(units_sold, axis=1)

median = np.median(units_sold, axis=1)

std = np.std(units_sold, axis=1)

variance = np.var(units_sold, axis=1)

items = ["Rice", "Atta", "Sugar", "Tea", "Oil", "Dal"]

print("Item   Total   Mean   Median   Std Dev   Variance")

for i in range(6):
    print(items[i], total[i], mean[i], median[i], std[i], variance[i])

daily_total = np.sum(units_sold, axis=0)

best_day = np.argmax(daily_total) + 1

print("Daily Total Sales:")
print(daily_total)

print("Highest Sales Day:", best_day)
```
