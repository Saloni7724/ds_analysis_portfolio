```python
"""
Task-1
"""

import numpy as np

print("NumPy Version:", np.__version__)

np.random.seed(42)

units_sold = np.random.randint(5, 60, size=(6, 7))

price_inr = np.array([52, 44, 45, 380, 140, 120])

print("Units Sold:")
print(units_sold)

print("Price INR:")
print(price_inr)

print("\nUnits Sold Details:")
print("Shape:", units_sold.shape)
print("Dimensions:", units_sold.ndim)
print("Size:", units_sold.size)
print("Data Type:", units_sold.dtype)
print("Item Size:", units_sold.itemsize)

print("\nPrice Details:")
print("Shape:", price_inr.shape)
print("Dimensions:", price_inr.ndim)
print("Size:", price_inr.size)
print("Data Type:", price_inr.dtype)
print("Item Size:", price_inr.itemsize)
```

```python
"""  Task-2
"""

stock_levels = np.full((6, 7), 100)

returns = np.zeros((6, 7))

in_stock = np.ones(6, dtype=bool)

item_ids = np.arange(101, 107)

day_marks = np.linspace(0, 6, 7)

item_pairs = np.eye(6)

print("Stock Levels:")
print(stock_levels)

print("Returns:")
print(returns)

print("In Stock:")
print(in_stock)

print("Item IDs:")
print(item_ids)

print("Day Marks:")
print(day_marks)

print("Item Pairs:")
print(item_pairs)

print("\nArray Details:")

arrays = [stock_levels, returns, in_stock,
          item_ids, day_marks, item_pairs]

for arr in arrays:
    print("Data Type:", arr.dtype)
    print("Memory:", arr.nbytes, "bytes")
```

```python
"""  Task-3
"""
import sys

bills_list = list(range(1000000))

bills_arr = np.array(bills_list)

bills_list_back = bills_arr.tolist()

print("List Size:", len(bills_list))
print("Array Size:", bills_arr.size)

print("List pointers only:", sys.getsizeof(bills_list))
print("Array Memory:", bills_arr.nbytes)

print("List Sum:", sum(bills_list))
print("Array Sum:", bills_arr.sum())

print("C Contiguous:", bills_arr.flags['C_CONTIGUOUS'])

print("Strides:", bills_arr.strides)
```

```python
""" Task-4
"""
def array_report(name, arr):

    print("Name:", name,
          "| Shape:", arr.shape,
          "| Dimensions:", arr.ndim,
          "| Size:", arr.size,
          "| Data Type:", arr.dtype,
          "| Item Size:", arr.itemsize,
          "| Memory:", arr.nbytes)

array_report("units_sold", units_sold)

array_report("price_inr", price_inr)

array_report("stock_levels", stock_levels)

array_report("returns", returns)

array_report("in_stock", in_stock)

array_report("item_ids", item_ids)

array_report("day_marks", day_marks)

array_report("item_pairs", item_pairs)

array_report("bills_arr", bills_arr)
```
