```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from pathlib import Path
from ydata_profiling import ProfileReport
import sweetviz as sv
import dtale

# Project folders
BASE = Path("..")
REPORTS = BASE / "reports"
PROCESSED = BASE / "data" / "processed"

REPORTS.mkdir(parents=True, exist_ok=True)
PROCESSED.mkdir(parents=True, exist_ok=True)

# Load seaborn taxi dataset
taxi = sns.load_dataset("taxis")

# Take 3000 random rows
taxi_sample = taxi.sample(n=3000, random_state=42)

# Reset index
taxi_sample = taxi_sample.reset_index(drop=True)

print("Taxi sample shape:", taxi_sample.shape)
print(taxi_sample.head())

# Save processed sample
taxi_sample.to_csv(PROCESSED / "taxi_sample.csv", index=False)
```

```python
profile = ProfileReport(
    taxi_sample,
    title="Taxi Trips Profile",
    explorative=True
)

profile.to_file(REPORTS / "taxi_profile.html")

print("YData Profiling report saved successfully!")
```

```python
# Force tip to be treated as numeric in both groups
feature_config = sv.FeatureConfig(
    force_num=["tip"]
)

# Compare cash trips against other payment types
report = sv.compare_intra(
    taxi_sample,
    taxi_sample["payment"] == "cash",
    ["Cash", "Other"],
    target_feat="fare",
    feat_cfg=feature_config
)

# Save Sweetviz report
report.show_html(
    filepath=str(REPORTS / "taxi_sweetviz_cash_vs_other.html"),
    open_browser=False
)

print("Sweetviz report saved successfully!")
```

```python
# Start D-Tale
d = dtale.show(taxi_sample)

# Print D-Tale link
print("D-Tale link:")
print(d.main_url())

# Open D-Tale in browser
d.open_browser()
```

```python
# Select numeric columns
numeric_df = taxi_sample.select_dtypes(include="number")

# Calculate correlation matrix
corr = numeric_df.corr()

print("Correlation Matrix:")
print(corr)

# Find strongest pair, excluding diagonal and pairs containing total
corr_check = corr.copy()

for col in corr_check.columns:
    corr_check.loc[col, col] = np.nan

if "total" in corr_check.columns:
    corr_check.loc["total", :] = np.nan
    corr_check.loc[:, "total"] = np.nan

# Convert correlation matrix to absolute values
absolute_corr = corr_check.abs()

# Find strongest pair
strongest_pair = absolute_corr.stack().idxmax()
strongest_value = corr.loc[strongest_pair[0], strongest_pair[1]]

print("\nStrongest relationship:")
print("Column 1:", strongest_pair[0])
print("Column 2:", strongest_pair[1])
print("Correlation:", strongest_value)
print("Absolute correlation:", abs(strongest_value))
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Taxi Dataset Correlation Heatmap")
plt.tight_layout()

plt.savefig(REPORTS / "taxi_corr_check.png")
plt.show()

print("Correlation heatmap saved successfully!")
```
